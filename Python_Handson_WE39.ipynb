```python
class remote:
    def power_button(self,onoff):
        if onoff:
            print("TV is switching on")
        else:
            print("TV is switching off")
    def sound_button(self,up):
        if up:
            print("sound is increasing")
        else:
            print("sound is decreasing")
```

# Higher Order Function
- A function that takes another function as an argument (or) 
- A function that returns another function

```python
lam_func1=lambda x:x.lower()
lam_func2=lambda x:x.upper()
def generate_mail_lambda(fname,lname,domain,lam_func_arg): # for repeated and common usage
    dom=lam_func_arg(domain)
    return fname+lname+dom
print(generate_mail_lambda('md','irfan','@INCEPTEz.com',lam_func1))
print(generate_mail_lambda('md','irfan','@INCEPTEz.com',lam_func2))
print(generate_mail_lambda('md','venkat','@INCEPTEz.com',lam_func2))
print(generate_mail_lambda('md','santhosh','@INCEPTEz.com',lam_func2))
print(generate_mail_lambda('md','jesus','@INCEPTEz.com',lam_func1))
print(generate_mail_lambda('md','allah','@INCEPTEz.com',lam_func2))
print(generate_mail_lambda('md','obama','@INCEPTEz.com',lam_func1))
```

# Nested Function

```python
def sal_hike(sal,hike):
    salhike=sal+hike 
    def incentives(incentive):  # CLOSURE FUnction
        return salhike+incentive  # Nested Function
    return incentives  # HOF - a function returned from another function
```

```python
incenfunc=sal_hike(10000,1000)
incenfunc
```

```python
incenfunc(500)
```
